## Asking Chatty how it works, it came up with:

- Known structure with unknown small part.
Example: you know most of a message m except a small unknown r, and ciphertext $c = (m+r)^e \bmod N$. Set
$f(x) = (m + x)^e - c \pmod N$,
then any small r satisfying the congruence is a root of $f(x) mod N$.

- Recovering low bits of a secret or small factor offsets.
You model the unknown offset x so the congruence becomes $f(x) \equiv 0 \pmod N$ and choose $f$ to be a polynomial in $x$ with integer coefficients (lifted to $\mathbb{Z}[x]$). The rest of the method uses shifts $x^j f(x) N^i$ to form the lattice.

General recipe for $f$:
	•	Express your unknown as a small integer x so the cryptographic relation becomes $f(x)\equiv0\pmod N$.
	•	Work in $\mathbb Z[x]$ (or $\mathbb Z_N[x]$ in Sage) and pick a reasonable bound X such that $|x|<X$.
	•	The method then builds the lattice of coefficient vectors of polynomials like $x^j N^i f(x)$ for chosen ranges of $i,j$. Parameter choices (how many shifts, m,t, etc.) trade off lattice dimension vs. guaranteed bounds; see Howgrave-Graham and Coppersmith for parameter heuristics.  ￼




$c = (\textit{Int}(\text{openECSC\{mind\_the\_gap\_}) + r)^3$

$f(x) = (m+x)^3 - c \pmod N$

```python
from sage.all import *
```

```python
m = int.from_bytes(prefix.encode("ascii"))
k = len(prefix.encode("ascii"))


K = Zmod(N)
R = PolynomialRing(K, 'x')
x = R.gen()
```

```python
for length in range(1,60):
    f = (m * 2**(8 * (k + length)) + x)**e - c

    roots = f.small_roots()
    if roots:
        print(len(roots),":",length,":",end="")
        print(roots[0].to_bytes())
```
