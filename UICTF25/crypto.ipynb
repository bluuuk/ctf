````python
import z3
from Crypto.Cipher import AES
import string
````

````python
a,b,c,d = [z3.Int(name) for name in "abcd"]
upper = z3.IntVal(2e4)
flag = bytes.fromhex("41593455378fed8c3bd344827a193bde7ec2044a3f7a3ca6fb77448e9de55155")
````

````python
def checksolution(a,b,c,d):
    try:
        test = AES.new( f"{a*b*c*d}".zfill(16).encode() , AES.MODE_ECB).decrypt(flag).decode("ASCII")
        print(test)
    except UnicodeDecodeError:
        pass
    
    try:
        test = AES.new( f"{-a*b*c*d}".zfill(16).encode() , AES.MODE_ECB).decrypt(flag).decode("ASCII")
        print(test)
    except UnicodeDecodeError:
        pass
    
checksolution(0,1,2,3)
````

````python
s = z3.Solver() 

s.add(
    [
        a >= b, 
        c >= d,
        a >= 0,
        b >= 0,
        c >= 0,
        d >= 0,
        a < upper,
        b < upper,
        d < upper,
        d < upper,
        a**4 + b**4 == c**4 + d**4 + 17
    ]
)

while s.check() == z3.sat:
    print(s.model())
    checksolution(
        s.model()[a].as_long(),
        s.model()[b].as_long(),
        s.model()[c].as_long(),
        s.model()[d].as_long(),
    )
    s.add(z3.Or(a != s.model()[a], b != s.model()[b],c != s.model()[c],d != s.model()[d]))
````
